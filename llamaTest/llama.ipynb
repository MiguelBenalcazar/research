```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch
import os
import gc
```

```python
# Check if CUDA is available
cuda_available = torch.cuda.is_available()
print(f"CUDA Available: {cuda_available}")

if cuda_available:
    print(f"CUDA Version: {torch.version.cuda}")
    print(f"Number of GPUs: {torch.cuda.device_count()}")
    print(f"Current GPU: {torch.cuda.get_device_name(0)}")
else:
    print("No GPU detected. Running on CPU.")
```

```python
model_path = os.path.join(os.path.dirname(os.getcwd()),"Models","Llama3.2-1B-Instruct-hf")
# Verify the path exists
if not os.path.exists(model_path):
    raise FileNotFoundError(f"The directory {model_path} does not exist. Please check the path.")
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_path, local_files_only=True)
```

```python
# user_prompt = ("A 40-year-old woman undergoing chemotherapy for ovarian carcinoma has developed progressive bilateral sensorineural hearing loss. Which chemotherapy drug is responsible for this side effect?"
# )

user_prompt = (
    "DMPA-True:a) Failure @ 0.3/100 WYb) 150 mg/3 monthly deliveredc) Weight gaind) Glucose intolerance occure) Anemia improves"
    "A. bcde"
    "B. abcd"
    "C. abce"
    "D. abde"
)
    


full_prompt = f"<|user|>{user_prompt}<|assistant|>"

# # Tokenize
# inputs = tokenizer(full_prompt, return_tensors="pt").to(model.device)
```

```python

model = AutoModelForCausalLM.from_pretrained(
    model_path,
    torch_dtype=torch.bfloat16,  # or torch.float16 if not supported
    device_map="cuda:0",
    local_files_only=True,
    
)

pipe = pipeline("text-generation", model=model, tokenizer=tokenizer)
output = pipe(full_prompt, truncation=True,  max_length=512,
    temperature=0.1,
    top_p=0.9,
    do_sample=True)

```

```python
print(output[0]["generated_text"])
torch.cuda.empty_cache()
```

```python
# del model_path, tokenizer, text, model, pipe, output
gc.collect()
torch.cuda.empty_cache()
torch.cuda.ipc_collect()
```

```python
text = "Puedes hablar espanol?"

print("ENCODE --> ", tokenizer.encode(text))
print("DECODE --> ",tokenizer.decode(tokenizer.encode(text)))

pipe = pipeline("text-generation", model=model, tokenizer=tokenizer)
output = pipe(text, truncation=True,  max_length=250)
```
